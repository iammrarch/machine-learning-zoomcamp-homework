### Q7 Sum of weights

```python
#All cars from Asia
asia_cars = df[
    df["origin"] == "Asia"
]
print(asia_cars)
```

```python
#Only columns vehicle_weight and model_year
vw_my = asia_cars[
    ["vehicle_weight","model_year"]
    ]
print(vw_my)
```

```python
#First 7 values
vw_my.head(n=7)
```

```python
#Underlying NumPy array
X = vw_my.head(n=7).values
print(X)
```

```python
#Matrix-matrix multiplication between the transpose of X and X
XTX = X.T.dot(X)
print(XTX)
```

```python
#Inverse of XTX
XTX_inv = np.linalg.pinv(XTX)
print(XTX_inv)
```

```python
#Creating array y
y = np.array([1100,1300,800,900,1000,1100,1200])
print(y)
```

```python
#Product of inverse of XTX and X.T
w = np.dot(np.dot(XTX_inv,X.T),y)
print(w)
```

```python
#sum of all the elements of w
w.sum()
```
