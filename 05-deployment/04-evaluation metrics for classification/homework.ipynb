### Evaluation Metrics for Classification

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
```

#### Data Preparation

***Data is from the 2026 lead-scoring dataset for classification***

```python
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/course_lead_scoring_2026.csv')
```

```python
#checking missing values
df.isnull().sum()
```

```python
#checking data types
df.dtypes
```

```python
#fixing missing values

missing_str = list(df.dtypes[df.dtypes == 'str'].index)
missing_float = list(df.dtypes[df.dtypes == 'float64'].index)

for col in missing_str:
    df[col] = df[col].fillna("NA")

for col in missing_float:
    df[col] = df[col].fillna("0.0")

df.isnull().sum()
```

```python
df.head()
```

```python
#splitting the dataset

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_train = df_train.converted.values
y_val = df_val.converted.values
y_test = df_test.converted.values

del df_train['converted']
del df_val['converted']
del df_test['converted']

len(df_train), len(df_val), len(df_test)
```

```python
df.columns
```

```python
#This is necessary for use in getting a subset of data from a 'boolean condition' rather than an object
categorical = ['lead_source', 'industry', 'employment_status', 'location']
numerical = ['annual_income', 'number_of_courses_viewed', 'interaction_count', 'lead_score',] #note that the target variable is not included
features = categorical + numerical
```

### Question 1

***Evaluating feature importance of numerical variables using ROC AUC***

```python
from sklearn.metrics import roc_auc_score

scores = {}

for col in numerical:
    auc = roc_auc_score(y_train, df_train[col])

    if auc < 0.5:
        auc = roc_auc_score(y_train, -df_train[col])

    scores[col] = auc

sorted(scores.items(), key=lambda x: x[1], reverse=True)
```

### Question 2

***Training the model***

```python
# Convert dataframes to dictionaries
train_dict = df_train[features].to_dict(orient='records')
val_dict = df_val[features].to_dict(orient='records')
```

```python
# One-hot encoding
dv = DictVectorizer(sparse=False)

X_train = dv.fit_transform(train_dict)
X_val = dv.transform(val_dict)
```

```python
# Train logistic regression
model = LogisticRegression(
    solver='liblinear',
    C=1.0,
    max_iter=1000
)

model.fit(X_train, y_train)
```

```python
# Predict probabilities
y_pred = model.predict_proba(X_val)[:, 1]
```

```python
# Compute AUC
auc = roc_auc_score(y_val, y_pred)

print(round(auc, 3))
```

### Question 3

***Precision and Recall***

```python
from sklearn.metrics import precision_score, recall_score

thresholds = np.arange(0.0, 1.01, 0.01)

scores = []

for t in thresholds:
    y_pred_t = (y_pred >= t).astype(int)

    p = precision_score(y_val, y_pred_t, zero_division=0)
    r = recall_score(y_val, y_pred_t, zero_division=0)

    scores.append((t, p, r))
```

***Plotting***

```python
#Creating DataFrame and plotting
df_scores = pd.DataFrame(
    scores,
    columns=['threshold', 'precision', 'recall']
)

plt.plot(df_scores.threshold, df_scores.precision, label='Precision')
plt.plot(df_scores.threshold, df_scores.recall, label='Recall')

plt.legend()
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.show()
```

***Finding the intersection***

```python
#Computing the absolute difference between precision and recall
df_scores['diff'] = (
    df_scores['precision'] - df_scores['recall']
).abs()
```

```python
#Ignoring thresholds where both precision and recall are zero
df_valid = df_scores[
    ~(
        (df_scores.precision == 0) &
        (df_scores.recall == 0)
    )
]
```

```python
#Finding the threshold with the smallest difference
best = df_valid.loc[df_valid['diff'].idxmin()]

print(best[['threshold', 'precision', 'recall', 'diff']])
```

### Question 4

***F1 score***

```python
#Computing F1 for all thresholds from 0.0 to 1.0 with increment 0.01

thresholds = np.arange(0.0, 1.01, 0.01)

scores = []

for t in thresholds:
    y_pred_t = (y_pred >= t).astype(int)

    p = precision_score(y_val, y_pred_t, zero_division=0)
    r = recall_score(y_val, y_pred_t, zero_division=0)

    if p + r == 0:
        f1 = 0
    else:
        f1 = 2 * p * r / (p + r)

    scores.append((t, p, r, f1))
```

```python
#Creating dataframe
df_scores = pd.DataFrame(
    scores,
    columns=['threshold', 'precision', 'recall', 'f1']
)

df_scores.sort_values('f1', ascending=False).head()
```

### Question 5

***5-Fold Cross-Validation***

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=5, shuffle=True, random_state=1)

scores = []

for train_idx, val_idx in kf.split(df_full_train):

    df_train_fold = df_full_train.iloc[train_idx]
    df_val_fold = df_full_train.iloc[val_idx]

    y_train = df_train_fold.converted.values
    y_val = df_val_fold.converted.values

    del df_train_fold['converted']
    del df_val_fold['converted']

    train_dict = df_train_fold[features].to_dict(orient='records')
    val_dict = df_val_fold[features].to_dict(orient='records')

    dv = DictVectorizer(sparse=False)

    X_train = dv.fit_transform(train_dict)
    X_val = dv.transform(val_dict)

    model = LogisticRegression(
        solver='liblinear',
        C=1.0,
        max_iter=1000
    )

    model.fit(X_train, y_train)

    y_pred = model.predict_proba(X_val)[:, 1]

    auc = roc_auc_score(y_val, y_pred)

    scores.append(auc)

print('AUC scores:', scores)
print('Mean:', np.mean(scores))
print('Std:', round(np.std(scores), 3))
```

### Question 6

***Hyperparameter Tuning***

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=5, shuffle=True, random_state=1)

for C in [0.000001, 0.001, 1]:
    scores = []

    for train_idx, val_idx in kf.split(df_full_train):
    
        df_train_fold = df_full_train.iloc[train_idx]
        df_val_fold = df_full_train.iloc[val_idx]
    
        y_train = df_train_fold.converted.values
        y_val = df_val_fold.converted.values
    
        del df_train_fold['converted']
        del df_val_fold['converted']
    
        train_dict = df_train_fold[features].to_dict(orient='records')
        val_dict = df_val_fold[features].to_dict(orient='records')
    
        dv = DictVectorizer(sparse=False)
    
        X_train = dv.fit_transform(train_dict)
        X_val = dv.transform(val_dict)
    
        model = LogisticRegression(
            solver='liblinear',
            C=C,
            max_iter=1000
        )
    
        model.fit(X_train, y_train)
    
        y_pred = model.predict_proba(X_val)[:, 1]
    
        auc = roc_auc_score(y_val, y_pred)
    
        scores.append(auc)
    
    print(
        f'C={C}, '
        f'mean={np.mean(scores):.3f}, '
        f'std={np.std(scores):.3f}'
    )
```
